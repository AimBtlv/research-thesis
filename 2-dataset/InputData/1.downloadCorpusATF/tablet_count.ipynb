```python
def count_atf(file_path):
    text_count = 0
    with open (file_path, 'r', encoding='utf-8') as file:
        for line in file:
            if line.strip().startswith('&'):
                text_count +=1
    return text_count 

atf_file_path = "/Users/aima/Downloads/atfOBP_School.atf"
try:
    total_texts = count_atf(atf_file_path)
    print (f'total atf in file:{total_texts}')
except FileNotFoundError:
    print (f"Errro, file {atf_file_path} was not found ")

```

```python
import pandas as pd 
import re
ATF_FILE = "/Users/aima/Desktop/Practice/GitHub/research-thesis/2-dataset/InputData/1.downloadCorpusATF/atfOBP_School_700.atf"
CSV_METADATA = "/Users/aima/Desktop/Practice/GitHub/research-thesis/2-dataset/InputData/1.downloadCorpusATF/metaOBP_School.csv"
OUTPUT_CSV= "/Users/aima/Desktop/Practice/GitHub/research-thesis/2-dataset/InputData/1.downloadCorpusATF/extract_metadata.csv"
def extract_atf_id(atf_path):
    atf_ids = set()
    id_pattern = re.compile(r'^&[PQX\d]*(\d+)')
    with open (atf_path, 'r',encoding ='utf-8') as file:
        for line in file:
            clean_line = line.strip()
            match =id_pattern.match(clean_line)
            if match:
                atf_ids.add(match.group(1))
    return atf_ids
def extract_meta (csv_path,atf_ids,output_path):
    df = pd.read_csv(csv_path, dtype={'artifact_id':str})
    filtered_df = df[df['artifact_id'].isin(atf_ids)]
    filtered_df.to_csv(output_path, index= False, encoding ='utf-8')
    print (f"atf tablet: {len(atf_ids)}")
    print(f"succefully extracted and saved to new csv:{len(filtered_df)}")

try:
    atf_ids = extract_atf_id(ATF_FILE)
    extract_meta(CSV_METADATA,atf_ids,OUTPUT_CSV)
except FileNotFoundError as e:
    print(f"Error check the pathways: {e}")
except KeyError:
    print("Error: In csv was not found 'artifact_id'")
                

    
```
